### Parameters:

1. values: Columns to aggregate.
2. index: Columns to use as the new row index.
3. columns: Columns to use as the new column headers.
4. aggfunc: Aggregation functions like mean, sum, count etc. By default it is mean.
5. fill_value: Value to replace missing data.
6. margins: Whether to add totals, default is false.
7. dropna: Whether to exclude missing values from the DataFrame, default is True.

```python
import pandas as pd

data = {
    "region":  ["North","South","East","North","South","East",
                "North","South","East","North","South","East"],
    "month":   ["Jan","Jan","Jan","Feb","Feb","Feb",
                "Mar","Mar","Mar","Apr","Apr","Apr"],
    "product": ["Laptop","Phone","Tablet","Laptop","Phone","Tablet",
                "Laptop","Phone","Tablet","Laptop","Phone","Tablet"],
    "revenue": [2400,1600,900,1200,2400,1350,3600,800,1800,1200,3200,2700]
}

df = pd.DataFrame(data)
print(df)
```

#### Build a pivot table showing total revenue, with region as rows and month as columns.

```python
df.pivot_table(values="revenue", index="region", columns="month", aggfunc="sum")
```

```python
df.pivot_table(values="revenue", index="region", columns="product", aggfunc=["sum", "mean"], margins = True, fill_value=0)
```
